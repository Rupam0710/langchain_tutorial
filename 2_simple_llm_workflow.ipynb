```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_groq import ChatGroq
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
# Example: model = ChatGroq(model="mixtral-8x7b-32768", groq_api_key="YOUR_GROQ_API_KEY")
model = ChatGroq(
    model="llama-3.1-8b-instant",
    temperature=0
)
```

```python
#create a state
class LLMState(TypedDict):
    question: str
    answer: str
```

```python
def llm_qa(state: LLMState) -> LLMState:
    
    #extract the question from the state
    question = state['question']

    #form a prompt for the LLM
    prompt = f"Answer the following question: {question}"

    #ask the LLM to answer the question
    answer = model.invoke(prompt).content

    #update the state with the answer
    state['answer'] = answer # type: ignore

    return state
```

```python
#define a graph
graph = StateGraph(LLMState)

#define the node
graph.add_node('llm_qa',llm_qa)

#define the edges
graph.add_edge(START, 'llm_qa')
graph.add_edge('llm_qa', END)

#compile the workflow
workflow = graph.compile()
```

```python
#Execute the workflow
initial_state = {'question': "who will win the FIFA World Cup 2026?"}
final_state = workflow.invoke(initial_state) # type: ignore
print(final_state)
```
